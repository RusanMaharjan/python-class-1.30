```python
'''
Syntax of If/Else Logic
==========================
if(condition) {
    logic... / print()
}
'''
temperature = float(input('Enter temperature value: '))

if temperature >= 30:
    print('Its hot today.')
elif temperature >= 15 and temperature < 30:
    print('Its warm today.')
else:
    print('Its cold today.')
```

```python
# Write a program that calculates income tax based on an income variable using these rules: 
# • Income <= $20,000: Tax is 0% 
# • Income over $20,000 but <= $50,000: Tax is 10% of the amount over $20,000. 
# • Income over $50,000: Tax is $3,000 + 20% of the amount over $50,000. 

income = int(input('Enter your annual income: '))

if income >= 0 and income <= 20000:
    print('0% Tax.')
    print('Total Income:',income)
elif income > 20000 and income <= 50000:
    print('10% Tax')
    tax = income * 0.1
    rem_amount = income - tax
    print(f'Tax Amount: ${tax}.')
    print(f'Remaining Income: ${rem_amount}.')
elif income > 50000:
    print('20% Tax')
    tax = (income * 0.2) + 3000
    rem_amount = income - tax
    print(f'Tax Amount: ${tax}.')
    print(f'Remaining Income: ${rem_amount}.')
else:
    print('Invalid Income.')
```

## Login Authentication System

```python
username = 'abc@gmail.com'
password = '1234567890'

uname = input('Enter username: ')
pwd = input('Enter password: ')

if uname == username and pwd == password:
    print('Login Successfull.')
elif uname == username and pwd != password:
    print('Incorrect Password.')
elif uname != username and pwd == password:
    print('User not found.')
else:
    print('Login Failed.')
```

## Restaurant Order System

```python
'''
match variable:
    case '1':
        print(...)
    case '2':
        print(...)
    case _:
        print(...)
'''
```

```python
print('Welcome to Restaurant Order System')
print('1. Breakfast')
print('2. Snacks')
print('3. Lunch')
print('4. Dinner')
print('5. Dessert')

choice = int(input('Enter your order choice: '))

match choice:
    case 1:
        print('You have ordered breakfast.')
        print('1. Tea')
        print('2. Egg')
        print('3. Bread')

        bf_choice = int(input('Enter your breakfast choice: '))

        match bf_choice:
            case 1:
                print('You have ordered Tea.')
            case 2:
                print('You have ordered Egg.')
            case 3:
                print('You have ordered Bread.')
            case _:
                print('Invalid breakfast choice.')
    case 2:
        print('You have ordered snacks.')
    case 3:
        print('You have orderd lunch.')
    case 4:
        print('You have ordered dinner.')
    case 5:
        print('You have ordered dessert.')
    case _:
        print('Invalid choice.')
```

```python
for i in range(1, 11):
    if i == 5:
        continue
    print(i)
```

```python
print('Password must contain atleast 1 uppercase character.')
print('Password must contain atleast 1 lowercase character.')
print('Password must contain atleast 1 digit character.')

password = input('Enter your password: ')

is_upper = False
is_lower = False
is_digit = False
is_special = False
special_char = '!@#$%^&*()_+_+|/<>?:;{}[]'

for word in password:
    if word.isupper():
        is_upper = True
        
    if word.islower():
        is_lower = True
        
    if word.isdigit():
        is_digit=True

    if word in special_char:
        is_special=True

if is_upper and is_lower and is_digit and is_special:
    print('Login Successfull.')
else:
    if not is_upper:
        print('Password must contain atleast 1 uppercase character.')

    if not is_lower:
        print('Password must contain atleast 1 lowercase character.')

    if not is_digit:
        print('Password must contain atleast 1 digit character.')

    if not is_special:
        print('Password must contain atleast 1 special character.')
```

## While Loop

```python
'''
-> While loop executes until the specified condition becomes False.
while condition:
    logic.../print...

while True:
    logic.../print...
'''
```

```python
secret_number = 7
guess_number = 0 # 7

while guess_number != secret_number: # 7 != 7 -> False
    guess_number = int(input('Enter guess number: ')) # 7
    if guess_number == secret_number: # 7 == 7 -> True
        print('Correct.')
    else:
        print('Incorrect.')
```

```python
secret_number = 7

condition = True

while condition:#/True
    guess_number = int(input('Enter guess number: ')) # 7

    if guess_number == secret_number: # 7 == 7
        print('Correct.')
        # break
        condition = False
    else:
        print('Incorrect.')
```

```python
pin = 1234
balance = 50000
attempts = 3

while attempts != 0:
    print(f'Remaining Attempts: {attempts}')
    user_pin = int(input('Enter your pin number: '))
    
    if user_pin == pin:
        amount = int(input('Enter your withdraw balance: '))

        if amount < balance:
            if amount % 1000 == 0:
                print(f'Rs.{amount} has been withdrawn.')
                break
            else:
                print('Withdraw amount must be multiple of 1000.')
        else:
            print('Withdraw amount must be less than available balance.')
    else:
        print('Incorrect pin number.')
        attempts -= 1

    if attempts == 0:
        print('ATM Blocked. Kindly visit nearest branch.')
        break
```
