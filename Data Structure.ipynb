```python
# The "Price Is Right" (Conditional Logic) You are given a list of product 
# prices: prices = [100, 250, 75, 400, 120, 310]. 
# 1. Write a program that creates two new empty lists: cheap_items and 
# expensive_items. 
# 2. Loop through the prices list. If a price is greater than 200, append it to 
# expensive_items. Otherwise, append it to cheap_items. 
# 3. Print both lists and calculate the total count of items in each using the len() 
# function.
```

```python
prices = [100, 250, 75, 400, 120, 310]
cheap_items = []
expensive_items = []
```

```python
for price in prices:
    if price > 200:
        expensive_items.append(price)
    else:
        cheap_items.append(price)
```

```python
print(expensive_items)
print(f'Total prices: {expensive_items}')
```

```python
print(cheap_items)
print(f'Total prices: {cheap_items}')
```

## Extract Unique and Duplicate values from list

```python
values = [1, 2, 3, 1, 2, 4, 5, 3]
unique = []
duplicate = []
```

```python
for val in values:
    if val in unique:
        if val not in duplicate:
            duplicate.append(val)
    else:
        unique.append(val)
```

```python
# A commercial parking garage monitors timestamps via a list of daily vehicle 
# durations in minutes: [45, 120, 310, 15, 640, 180]. 

# The pricing rules dictate that any duration under 60 minutes is charged at $5.
# Any duration between 60 and 240 minutes is billed at $12.
# Anything exceeding 240 minutes is flag-rated as an all-day fee of $30. 
```

```python
minutes = [45, 120, 310, 15, 640, 180]

low_charge = []
average_charge = []
high_charge = []

for minute in minutes:
    if minute <= 60:
        low_charge.append(minute)
        print(f'You parking time {minute} min and your charge is: $5.')
    elif minute > 60 and minute < 240:
        average_charge.append(minute)
        print(f'You parking time {minute} min and your charge is: $12.')
    else:
        high_charge.append(minute)
        print(f'You parking time {minute} min and your charge is: $30.')
```

```python
print(low_charge)
print(average_charge)
print(high_charge)
```

```python
# Core Concept: Data Structures (Lists, Dictionaries, and Mutability) 
# • Difficulty: Basic 
# • Scenario: A small electronics warehouse tracks its current stock items inside a Python 
# dictionary, where keys are product IDs and values are dictionaries containing both item 
# name and stock quantities (e.g., {'PROD101': {'name': 'Wireless Mouse', 'stock': 12}}). 
# • Task: Write a script that checks this tracking dictionary. If an item's stock count drops 
# below 15 units, append its name to a list called low_stock_warnings. If an item has exactly 
# 0 units left, move that product ID out of the inventory dictionary entirely and place it into 
# a separate list named out_of_stock_discontinued.
```

```python
products = {
    'PROD101': {
        'name': 'Wireless Mouse',
        'stock': 12
    },
    'PROD102': {
        'name': 'Laptop',
        'stock': 0
    },
    'PROD103': {
        'name': 'Mobile',
        'stock': 50
    },
    'PROD104': {
        'name': 'Watch',
        'stock': 0
    }
}

low_stock_warnings = [] # if stock < 15
out_of_stock_discontinued = [] # if stock value = 0 then append ID

products
```

```python
for key, value in list(products.items()):
    quantity = value['stock']
    name = value['name']
    pro_id = key

    if quantity <= 0:
        out_of_stock_discontinued.append(pro_id)
        del products[pro_id]
        
    elif quantity > 0 and quantity <= 15:
        low_stock_warnings.append(name)

print(f'Low Stock Warning: {low_stock_warnings}')
print(f'Out of Stock Discontinued: {out_of_stock_discontinued}')
print(f'Products: {products}')
```
