```python
# If - elif - else
# Temperature
temperature = float(input('Enter temperature value: ')) # 5

if temperature >= 28: # 5 >= 28 -> False
    print('Its hot today.')
elif temperature >= 15 and temperature < 28: # 5 >= 15 -> False and 5 < 28 -> True -> False
    print('Its warm today.')
else:
    print('Its cold today.')
```

```python
'''
- 80 - 100 -> Grade: A
- 60 - 79 -> Grade: B
- 50 - 59 -> Grade: C
- 40 - 49 -> Grade: D
-> Fail
'''
```

```python
grade = float(input('Enter your grade: '))

if grade >= 80 and grade <= 100:
    print('Grade: A')
elif grade >= 60 and grade < 80:
    print('Grade: B')
elif grade >= 50 and grade < 60:
    print('Grade: C')
elif grade >= 40 and grade < 50:
    print('Grade: D')
else:
    print('Fail')
```

```python
username = 'abc@gmail.com'
pwd = '1234567890'

email = input('Enter your email: ')
password = input('Enter your password: ')

if email == username and password == pwd:
    print('Login Successfull.')
elif email == username and password != pwd:
    print('Incorrect password.')
elif email != username and password == pwd:
    print('User not found.')
else:
    print('Login Failed.')
```

```python
# Nested if
username = 'abc@gmail.com'
pwd = '1234567890'

email = input('Enter your email: ')
password = input('Enter your password: ')

if email != username and password != pwd:
    print('email or password invalid.')
else:
    if email == username:
        if password == pwd:
            print('Login Successfull.')
        else:
            print('Incorrect Password.')
    else:
        print('User not found.')
```

```python
'''
- Pin -> Enter pin number
- Balance -> withdraw balance must be less than available balance.
- Balance -> we only can withdraw balance of 1000s.
'''
```

```python
pin = 1234
balance = 50000

user_pin = int(input('Enter your pin number: '))
if user_pin == pin:
    amount = int(input('Enter your withdraw amount: '))
    if amount < balance:
        if amount % 1000 == 0:
            print(f'Rs.{amount} has been withdrawn from A/C no. 1234567890.')
        else:
            print('Error: Amount must be multiple of 1000s.')
    else:
        print('Error: Withdraw amount must be less than available balance.')
else:
    print('Error: Incorrect Pin Number.')
```
