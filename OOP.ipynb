### Single Inheritance

```python
class Customers:
    def __init__(self, customer_id, customer_name, customer_address, customer_phone, customer_type):
        self.customer_id = customer_id
        self.customer_name = customer_name
        self.customer_address = customer_address
        self.customer_phone = customer_phone
        self.customer_type = customer_type

    def customer_details(self):
        print(f'Customer Id: {self.customer_id}')
        print(f'Customer Name: {self.customer_name}')
        print(f'Customer Address: {self.customer_address}')
        print(f'Customer Phone: {self.customer_phone}')
        print(f'Customer Type: {self.customer_type}')
```

```python
class Orders(Customers): # order class inherits customer class
    def __init__(self, order_id, order_status, order_date, order_destination,
                customer_id, customer_name, customer_address, customer_phone, customer_type):
        super().__init__(customer_id, customer_name, customer_address, customer_phone, customer_type)
        self.order_id = order_id
        self.order_status = order_status
        self.order_date = order_date
        self.order_destination = order_destination

    def order_details(self):
        print(f'Order Id: {self.order_id}')
        print(f'Order Status: {self.order_status}')
        print(f'Order Date: {self.order_date}')
        print(f'Order Destination: {self.order_destination}')
```

### Multi-Level Inheritance

```python
'''
- It requires 3 or more than 3 classes.
- Class C Inherits Class B Inherits Class A
'''
```

```python
# Grand Parent
class Customers:
    def __init__(self, customer_id, customer_name, customer_address, customer_phone, customer_type):
        self.customer_id = customer_id
        self.customer_name = customer_name
        self.customer_address = customer_address
        self.customer_phone = customer_phone
        self.customer_type = customer_type

    def customer_details(self):
        print(f'Customer Id: {self.customer_id}')
        print(f'Customer Name: {self.customer_name}')
        print(f'Customer Address: {self.customer_address}')
        print(f'Customer Phone: {self.customer_phone}')
        print(f'Customer Type: {self.customer_type}')
```

```python
# Parent Class
class Orders(Customers): # order class inherits customer class
    def __init__(self, order_id, order_status, order_date, order_destination,
                customer_id, customer_name, customer_address, customer_phone, customer_type):
        super().__init__(customer_id, customer_name, customer_address, customer_phone, customer_type)
        self.order_id = order_id
        self.order_status = order_status
        self.order_date = order_date
        self.order_destination = order_destination

    def order_details(self):
        print(f'Order Id: {self.order_id}')
        print(f'Order Status: {self.order_status}')
        print(f'Order Date: {self.order_date}')
        print(f'Order Destination: {self.order_destination}')
```

```python
# Child Class
class Order_Items(Orders):
    def __init__(self, item_id, item_price, item_quantity, item_discount,
                order_id, order_status, order_date, order_destination,
                customer_id, customer_name, customer_address, customer_phone, customer_type): # get value from user
        # Super Constructor Inheritance
        super().__init__(order_id, order_status, order_date, order_destination,
                customer_id, customer_name, customer_address, customer_phone, customer_type) # set value to specific classes
        self.item_id = item_id
        self.item_price = item_price
        self.item_quantity = item_quantity
        self.item_discount = item_discount

    def display_items(self):
        print(f'Item Id: {self.item_id}')
        print(f'Item Price: Rs.{self.item_price}')
        print(f'Item Quantity: {self.item_quantity}')
        print(f'Item Discount: {self.item_discount}')
```

### Multiple Inheritance

```python
class Customers:
    def __init__(self, customer_id, customer_name, customer_address, customer_phone, customer_type):
        self.customer_id = customer_id
        self.customer_name = customer_name
        self.customer_address = customer_address
        self.customer_phone = customer_phone
        self.customer_type = customer_type

    def customer_details(self):
        print(f'Customer Id: {self.customer_id}')
        print(f'Customer Name: {self.customer_name}')
        print(f'Customer Address: {self.customer_address}')
        print(f'Customer Phone: {self.customer_phone}')
        print(f'Customer Type: {self.customer_type}')
```

```python
class Staff:
    def __init__(self, staff_id, staff_name, staff_type, staff_position, staff_contact, staff_email):
        self.staff_id = staff_id
        self.staff_name = staff_name
        self.staff_type = staff_type
        self.staff_position = staff_position
        self.staff_contact = staff_contact
        self.staff_email = staff_email

    def staff_details(self):
        print(f'Staff Id: {self.staff_id}')
        print(f'Staff Name: {self.staff_name}')
        print(f'Staff Type: {self.staff_type}')
        print(f'Staff Position: {self.staff_position}')
        print(f'Staff Contact: {self.staff_contact}')
        print(f'Staff Email: {self.staff_email}')
```

```python
class Orders(Customers, Staff):
    def __init__(self, order_id, order_status, order_date, order_destination,
                customer_id, customer_name, customer_address, customer_phone, customer_type,
                staff_id, staff_name, staff_type, staff_position, staff_contact, staff_email):
        
        # Constructor of both staff and customer inherited
        Customers.__init__(self, customer_id, customer_name, customer_address, customer_phone, customer_type)
        Staff.__init__(self, staff_id, staff_name, staff_type, staff_position, staff_contact, staff_email)
        
        self.order_id = order_id
        self.order_status = order_status
        self.order_date = order_date
        self.order_destination = order_destination

    def order_details(self):
        print(f'Order Id: {self.order_id}')
        print(f'Order Status: {self.order_status}')
        print(f'Order Date: {self.order_date}')
        print(f'Order Destination: {self.order_destination}')
```

## Encapsulation

```python
'''
Access Modifiers
--------------------
1. Public
    - self.var_name = var_name
    - Variable data can be access by any object inside the class or outside the class
    
2. Private (__) double underscore
    - self.__var_name = var_name
    - Variable data cannot be accessed outside the class.
    - Data must be accesed using function.
    
3. Protected (_) single underscore
    - self._var_name = var_name
    - Variable data can be access inside the class and outside the class.
    - If we need to access data outside the class we have to use protected modifier symbol (_).
    - Data must be accessed using function.
'''
```

### Private Access Modifier

```python
class Customers:
    def __init__(self, customer_id, customer_name, customer_address, customer_phone, customer_type):
        self.__customer_id = customer_id
        self.__customer_name = customer_name
        self.__customer_address = customer_address
        self.__customer_phone = customer_phone
        self.__customer_type = customer_type

    def customer_details(self):
        print(f'Customer Id: {self.__customer_id}')
        print(f'Customer Name: {self.__customer_name}')
        print(f'Customer Address: {self.__customer_address}')
        print(f'Customer Phone: {self.__customer_phone}')
        print(f'Customer Type: {self.__customer_type}')
```

```python
class Orders(Customers): # order class inherits customer class
    def __init__(self, order_id, order_status, order_date, order_destination,
                customer_id, customer_name, customer_address, customer_phone, customer_type):
        super().__init__(customer_id, customer_name, customer_address, customer_phone, customer_type)
        self.__order_id = order_id
        self.__order_status = order_status
        self.__order_date = order_date
        self.__order_destination = order_destination

    def order_details(self):
        print(f'Order Id: {self.__order_id}')
        print(f'Order Status: {self.__order_status}')
        print(f'Order Date: {self.__order_date}')
        print(f'Order Destination: {self.__order_destination}')
```

## Polymorphism

```python
'''
- It will have multiple independent classes with same function name and arguments
- All classes object must be executed using external independent function.

- A function with many forms is Polymorphism.
'''
```

```python
class Esewa:
    def pay(self, send_from, send_to, amount):
        print(f'Rs.{amount} has been withdrawn using Esewa from {send_from} to {send_to}.')

class Khalti:
    def pay(self, send_from, send_to, amount):
        print(f'Rs.{amount} has been withdrawn using Khalti from {send_from} to {send_to}.')

class FonePay:
    def pay(self, send_from, send_to, amount):
        print(f'Rs.{amount} has been withdrawn using FonePay from {send_from} to {send_to}.')
```

```python
def payment(obj, send_from, send_to, amount):
    obj.pay(send_from, send_to, amount)
```

```python
'''
- Function Overloading - Same function, Same argument, Different Values
'''
```

```python
payment(Khalti(), 'Ram', 'Shyam', 500)
```

## Abstraction

```python
'''
- Abstract -> Data Hiding and Implementation Hiding
- To use abstraction concept parent class must be an abstract class.
- If we create an abstract class there must be atleast one abstract function.

- If normal class -> Inheritance
- If abstract class -> Implements
'''
```

```python
# Abstract Base Class
from abc import ABC, abstractmethod
```

```python
class MobileBanking(ABC):
    @abstractmethod
    def userDetails(self):
        pass
```

```python
class NabilBankMobileBanking(MobileBanking):
    def connection(self):
        print('Connected to Nabil Bank Database Server...')

    def userDetails(self):
        print(f'A/C No.: 1234567890\nBalance: Rs.60000')
```

```python
nb = NabilBankMobileBanking()
```

```python
nb.connection()
```

```python
nb.userDetails()
```
