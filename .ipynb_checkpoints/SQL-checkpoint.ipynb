```python
'''
    Database
    ------------
    1. RDBMS (Relational Database Mangement System)
        - MySQL
        - Posgres SQL (PgSQL)
        - MSSQL Server (Microsoft SQL Server)
        - Oracle
        
    2. DBMS (Database Management System)
        - SQLite, NoSQL, MongoDB, CouchDB, ....

    SQL (Structured Query Language)
    ---------------------------------
    - Database
        - Tables -> Entities
            - Columns -> Attributes
                - Rows -> Tuples
                    - Data -> Information
    
    Create Table
    --------------
    Keys
    --------
        1. Primary key
        2. Foreign Key
        
    Constraints
    ---------------
        1. Primary Key -> unique and not null
        2. Foreign key -> creates relation between table and validate data.
        3. Unique Key -> Unique + Null (Not Null)
        4. Null -> Default
        5. Not Null
        6. Check -> Employee Type (Senior, Junior)

    Concurrency (Database DeadLock)
    --------------------------------
    X Lock (Exclusive Lock) -> Create, Insert, Update, Delete
    S Lock (Shared Lock) -> Read (Select)

    Relationship
    -------------
    1. One to One
    2. One to Many
    3. Many to One
    4. Many to Many
'''
```

```python
import sqlite3
```

```python
conn = sqlite3.connect('ecommerce.db')
cursor = conn.cursor()
```

## Create Table

### Customer Table

```python
try:
    cursor.execute('''
        Create Table Customers (
        	customer_id integer primary key autoincrement,
        	customer_name text not null,
        	email text not null unique,
        	phone text not null unique,
        	address text not null,
        	customer_type text not null CHECK (customer_type  in (
        											'Regular', 'Non-Regular', 'VIP'
        										))
        );
    ''')
    print('Customer table created successfully.')
except sqlite3.OperationalError as e:
    print(f'Error: {e}')
```

### Staff Table

```python
try:
    cursor.execute('''
        Create Table Staffs (
            staff_id integer primary key autoincrement,
            staff_name text not null,
            staff_type text not null check (staff_type in ('Full-Time', 'Part-Time', 'Contract')),
            staff_position text not null check (staff_position in ('Intern', 'Junior', 'Senior')),
            staff_email text not null unique
        )
    ''')
    print('Staff table created successfully.')
except sqlite3.OperationalError as e:
    print(f'Error: {e}')
```

### Orders Table

```python
try:
    cursor.execute('''
        Create Table Orders (
            order_id integer primary key autoincrement,
            order_status text not null check (order_status in ('Completed', 'Processing', 'Pending', 'Rejected')),
            -- YYYY-MM-DD
            order_date date not null,
            order_desc text,
            customer_id integer not null,
            staff_id integer not null,
            foreign key (customer_id) references Customers(customer_id),
            foreign key (staff_id) references Staffs(staff_id)
        );
    ''')
    print('Staff table created successfully.')
except sqlite3.OperationalError as e:
    print(f'Error: {e}')
```

## Insert Data

### Customer Table

```python
try:
    cursor.execute('''
        Insert into Customers (customer_name, email, phone, address, customer_type)
        values 
        ('Sita', 'sita@gmail.com', '5432167890', 'Lalitpur', 'Regular'),
        ('Bob', 'bob@gmail.com', '1230987654', 'Bhaktapur', 'Regular'),
        ('Arthur', 'arthur@gmail.com', '0987123456', 'Makwanpur', 'Non-Regular');
    ''')
    conn.commit()
    print('Customer data insert successfully.')
except sqlite3.OperationalError as e:
    print(f'Error: {e}')
except sqlite3.IntegrityError as ie:
    print(f'Error: {ie}')
```

### Staff Table

```python
try:
    cursor.execute('''
        Insert into Staffs (staff_name, staff_type, staff_position, staff_email)
        values
        ('Dutch', 'Full-Time', 'Senior', 'dutch@gmail.com'),
        ('John', 'Part-Time', 'Junior', 'john@gmail.com'),
        ('Jacob', 'Contract', 'Intern', 'jacob@gmail.com');
    ''')
    conn.commit()
    print('Staff data inserted successfully.')
except sqlite3.OperationalError as e:
    print(f'Error: {e}')
except sqlite3.IntegrityError as ie:
    print(f'Error: {ie}')
```

### Order Table

```python
# order_status, order_date, order_desc, customer_id, staff_id
try:
    cursor.execute('''
        Insert into Orders (order_status, order_date, order_desc, customer_id, staff_id)
        Values
        ('Processing', '2026-09-21', 'Order 1', 2, 3),
        ('Completed', '2026-09-19', 'Order 2', 1, 3),
        ('Pending', '2026-09-22', 'Order 3', 4, 1);
    ''')
    conn.commit()
    print('Orders data inserted successfully.')
except sqlite3.OperationalError as e:
    print(f'Error: {e}')
except sqlite3.IntegrityError as ie:
    print(f'Error: {ie}')
```

## Update Table Data

### Orders Table

```python
try:
    cursor.execute('''
        Update Orders set order_status = 'Completed', order_date = '2026-09-22' where order_id = 1;
    ''')
    conn.commit()
    print('Orders table data updated.')
except sqlite3.OperationalError as e:
    print(f'Error: {e}')
except sqlite3.IntegrityError as ie:
    print(f'Error: {ie}')
```

## Select Data

### Customer Data

```python
try:
    cursor.execute('''
        select * from customers where address = 'Lalitpur';
    ''')
    data = cursor.fetchall()
    for row in data:
        print(row)
except sqlite3.OperationalError as e:
    print(e)
```

```python
try:
    cursor.execute('''
        select * from customers where customer_id > 1 and customer_id <= 4;
    ''')
    data = cursor.fetchall()
    for row in data:
        print(row)
except sqlite3.OperationalError as e:
    print(e)
```

```python
# values of between condition is inclusive.
try:
    cursor.execute('''
        select * from customers where customer_id between 2 and 4;
    ''')
    data = cursor.fetchall()
    for row in data:
        print(row)
except sqlite3.OperationalError as e:
    print(e)
```

```python
try:
    cursor.execute('''
        select * from customers where address = 'Kathmandu' or address = 'Lalitpur';
    ''')
    data = cursor.fetchall()
    for row in data:
        print(row)
except sqlite3.OperationalError as e:
    print(e)
```

```python
try:
    cursor.execute('''
        select * from customers where address in ('Kathmandu', 'Lalitpur');
    ''')
    data = cursor.fetchall()
    for row in data:
        print(row)
except sqlite3.OperationalError as e:
    print(e)
```

#### Like

```python
'''
wildcards
-----------
% -> pattern matching
_ -> single word
'''
```

```python
# Find customer name whose name starts with letter 'S'.
# Find customer name whose name ends with letter 'm'.
# Find customer name whose name must consist of only 3 letters.
```

```python
try:
    cursor.execute('''
        select * from customers where customer_name like 'S%';
    ''')
    data = cursor.fetchall()
    for row in data:
        print(row)
except sqlite3.OperationalError as e:
    print(e)
```

```python
try:
    cursor.execute('''
        select * from customers where customer_name like '%m';
    ''')
    data = cursor.fetchall()
    for row in data:
        print(row)
except sqlite3.OperationalError as e:
    print(e)
```

```python
try:
    cursor.execute('''
        select * from customers where customer_name like '___';
    ''')
    data = cursor.fetchall()
    for row in data:
        print(row)
except sqlite3.OperationalError as e:
    print(e)
```

```python
try:
    cursor.execute('''
        select * from customers where customer_name like '%a%';
    ''')
    data = cursor.fetchall()
    for row in data:
        print(row)
except sqlite3.OperationalError as e:
    print(e)
```
