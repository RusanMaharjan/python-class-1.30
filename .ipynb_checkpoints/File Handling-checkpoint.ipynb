## CSV File (Comma Separated Value)

### Create File

```python
import csv # package to work with csv file
products = [
    ['Product Name', 'Product Quantity', 'Product Price', 'Model Year'],
    ['Mobile', 2, 3000, 2022],
    ['Bag', 6, 5000, 2025]
]
```

```python
# open(file_name, operation, newline='')
with open('products.csv', 'w', newline='') as file:
    writer = csv.writer(file) # creates file
    writer.writerows(products) # inserts data into file
```

### Read File

```python
with open('products.csv', 'r') as file:
    reader = csv.reader(file) # extract rows of file into an object
    for row in reader: # loop object to extract data rows
        print(row)
```

### Append Data into File

```python
new_products = [
    ['Charger', 3, 1500, 2022],
    ['TV', 4, 10000, 2025]
]
```

```python
with open('products.csv', 'a', newline='') as file:
    writer = csv.writer(file)
    writer.writerows(new_products)
```

## Dictionary Data to CSV

```python
product_json = [
    {
        'Product Name': 'Charger',
        'Product Brand': 'Anker',
        'Product Price': '$150'
    },
    {
        'Product Name': 'Mobile',
        'Product Brand': 'Samsung',
        'Product Price': '$2000'
    },
    {
        'Product Name': 'Watch',
        'Product Brand': 'NavyForce',
        'Product Price': '$200'
    }
]
```

```python
# open(file_name, operation, newline='')

with open('products_dict.csv', 'w', newline='') as file:
    columns = ['Product Name', 'Product Brand', 'Product Price']
    writer = csv.DictWriter(file, fieldnames=columns)
    writer.writeheader()
    writer.writerows(product_json)
```

```python
# Read data
with open('products_dict.csv', 'r') as file:
    # parse file data to dictionary format
    reader = csv.DictReader(file)
    for row in reader:
        for key, val in row.items():
            print(f'{key}: {val}')
        print('\n')
```

```python
new_product_json = [
    {
        'Product Name': 'Laptop',
        'Product Brand': 'HP',
        'Product Price': '$1500'
    }
]
```

```python
with open('products_dict.csv', 'a', newline='') as file:
    columns = ['Product Name', 'Product Brand', 'Product Price']
    writer = csv.DictWriter(file, fieldnames=columns)
    writer.writerows(new_product_json)
```
