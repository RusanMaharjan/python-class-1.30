```python
'''
1. Try -> runnable code
2. Except (catch) -> get error from runnable code and handles it
3. Else -> any content that needs to be print.
4. Finally -> Prints content even if program executed successfully or not.
5. Raise -> Used with custom exception
'''
```

```python
a = int(input('Enter first number: '))
b = int(input('Enter second number: '))

c = a / b

print(c)
```

```python
try:
    a = int(input('Enter first number: '))
    b = int(input('Enter second number: '))
    
    c = a / b
    
except ValueError as ve:
    print(f'Error: Only number values are accepted in input fields.')
except ZeroDivisionError as zde:
    print(f'Error: Number cannot be divisible by zero.')
else:
    print(c)
finally:
    print('Program ended successfully.')
```

## Custom Exception

```python
# 1. Class
class AgeError(Exception):
    pass
```

```python
# 2. Function
def checkAge(age: int):
    if age >= 18:
        print('Person is eligible to vote.')
    else:
        raise AgeError(f'Person with age {age} is not eligible to vote.')
```

```python
try:
    age = int(input('Enter your age: '))
    checkAge(age)
except AgeError as ae:
    print(f'AgeError: {ae}')
except ValueError as ve:
    print(f'Error: Only number value is accepted in age input field.')
```

```python
class PinError(Exception):
    pass

class WithdrawAmountError(Exception):
    pass

class BalanceError(Exception):
    pass
```

```python
def atm():
    pin = 1234
    balance = 50000
    
    user_pin = int(input('Enter your pin number: '))
    
    if user_pin == pin: # 123 == 1234 -> False
        amount = int(input('Enter your withdraw balance: '))
    
        if amount < balance: # 5500 < 50000
            if amount % 1000 == 0: # 5500 % 1000 == 0
                print(f'Rs.{amount} has been withdrawn.')
            else:
                raise BalanceError('Withdraw amount must be multiple of 1000.')
        else:
            raise WithdrawAmountError('Withdraw amount must be less than available balance.')
    else:
        raise PinError('Incorrect pin number.')
```

```python
try:
    atm()
except PinError as pe:
    print(f'PinError: {pe}')
except WithdrawAmountError as wae:
    print(f'WithdrawAmountError: {wae}')
except BalanceError as be:
    print(f'BalanceError: {be}')
```
