# password must start with character and must end with #. Password must be 8 character long.

```python
'''
- re.sub() -> remove given pattern from data
- re.findall() -> fetch given pattern from data
- re.match() -> check if given data pattern is correct or not.
'''

import re # regular expression
# first letter must be a-z or A-Z
# remaining 6 letters can be any values
# last letter must be #
# $ -> to end pattern
def validatePassword(password):
    pattern = r'^[a-zA-Z].{6}#$'
    validate = bool(re.match(pattern, password))
    return 'Pattern matched' if validate else 'Unmatched Pattern'
```

```python
validatePassword('Afghjku#')
```

#### All Argument (*args)

```python
price1 = 30000
price2 = 40000
price3 = 70000
```

```python
def proudctDetails(*abc):
    print(abc)
```

```python
proudctDetails('Mobile', 'MI', 5000)
```

```python
def changeCurrency(prefix, *args):
    for data in args:
        print(f'{prefix}{data}')
```

```python
changeCurrency('Rs.', price1, price2, price3)
```

#### All Keyword Arguments (**kwargs)

```python
def kw_productDetails(**kwargs):
    print(kwargs)

    for key, val in kwargs.items():
        print(f'{key}: {val}')
```

```python
kw_productDetails(Product_Name='Mobile', Product_Price=3000)
```

## Lambda Function

```python
# One line function / arrow function
def add(a, b):
    return f'Addition of {a} and {b} is: {a + b}.'
```

```python
add(3, 5)
```

```python
'''
var_name = lambda arg1, arg2 : return value
'''
addition = lambda a, b : f'Addition of {a} and {b} is: {a + b}'
```

```python
addition(5, 6)
```

```python
# The "Price Is Right" (Conditional Logic) You are given a list of product 
# prices: prices = [100, 250, 75, 400, 120, 310]. 
# 1. Write a program that creates two new empty lists: cheap_items and 
# expensive_items. 
# 2. Loop through the prices list. If a price is greater than 200, append it to 
# expensive_items. Otherwise, append it to cheap_items. 
# 3. Print both lists and calculate the total count of items in each using the len() 
# function.
```

```python
prices = [100, 250, 75, 400, 120, 310]
cheap_items = []
expensive_items = []
```

```python
item_prices = lambda : [cheap_items.append(price) if price <= 200 else expensive_items.append(price) for price in prices]
```

```python
item_prices()
```

```python
print(cheap_items)
print(expensive_items)
```

```python
cheap_details = lambda : [price  for price in prices if price <= 200]
expensive_details = lambda : [price  for price in prices if price > 200]
```

```python
cheap_details()
```

```python
expensive_details()
```

## Nested Functions

```python
## calculation
def add(a, b):
    return f'Addition of {a} and {b} is: {a + b}'

def sub(a, b):
    return f'Subtraction of {a} and {b} is: {a - b}'

def mul(a, b):
    return f'Multiplication of {a} and {b} is: {a * b}'

def div(a, b):
    return f'Division of {a} and {b} is: {a / b}'
```

```python
def calculator():
    print('1. Addition')
    print('2. Subtraction')
    print('3. Multiplication')
    print('4. Division')

    choice = int(input('Enter your choice: '))

    x = int(input('Enter first number: '))
    y = int(input('Enter second number: '))
    
    if choice == 1:
        return add(x, y)
    elif choice == 2:
        return sub(x, y) 
    elif choice == 3:
        return mul(x, y)
    elif choice == 4:
        return div(x, y)
    else:
        print('Invalid choice.')
```

```python
calculator()
```
